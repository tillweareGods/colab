```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import normalize
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error,mean_squared_error
import numpy as np
```

```python
data = pd.read_csv("ParisHousing.csv")
data.head(n=10)
```

```python
my_colmn = data.columns
def null_check(the_data):
  for i in range(len(my_colmn)):
    output = the_data[my_colmn[i]]
    if output.isnull != True:
      print(my_colmn[i] , "is not null")
    elif output.isnull == True:
      print(my_colmn[i], "is null ")
```

```python
null_check(data)
```

```python
def normalize(the_data):
  for i in range(len(my_colmn) -1):
    the_data[my_colmn[i]] = the_data[my_colmn[i]]/the_data[my_colmn[i]].max()
```

```python
normalize(data)
```
