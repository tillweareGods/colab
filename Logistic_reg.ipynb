```python
def categorize_video_views(df):
    median_video_view = 1265499
    min_video_view = 23
    max_video_view = 7990989731

    def categorize(row):
        if row['Video Views'] < (min_video_view + 0.5 * (median_video_view - min_video_view)):
            return 1
        elif (min_video_view + 0.5 * (median_video_view - min_video_view)) <= row['Video Views'] <= (median_video_view + 0.5 * (max_video_view - median_video_view)):
            return 2
        else:
            return 3
################### two classifications only ######################################
    def categorize2(row):
        if row['Video Views'] < median_video_view:
            return 1
        else:
          return 2

    data['view_count'] = data.apply(categorize, axis=1)
    data['view_count_2'] = data.apply(categorize2, axis= 1)
    return df


```

```python
numeric_columns = ["Total Channel Subcribers", "Total Chanel Views"]
for col in numeric_columns:
    data[col] = pd.to_numeric(data[col], errors='coerce')

# Drop rows with NaN values in the specified columns
data = data.dropna(subset=numeric_columns)
```

```python
df = categorize_video_views(data)
```
